```python
import pandas as pd
import json
import os

path="C:\\Users\\yadun\\OneDrive\\Desktop\\Abhinand\\vscode\\pulse\\data\\aggregated\\transaction\\country\\india\\state"
Agg_state_list = os.listdir(path)

clm = {'State': [],'Year': [],'Quarter': [],'Transaction_type': [],'Transaction_count': [],'Transaction_amount': []}

for state in Agg_state_list:                                  
    state_path = os.path.join(path, state)
    year_list = os.listdir(state_path)

    for year in year_list:                                   
        year_path = os.path.join(state_path, year)
        quarter_list = os.listdir(year_path)

        for q in quarter_list:                                
            if q.endswith(".json"):
                file_path = os.path.join(year_path, q)
                with open(file_path, "r") as f:
                    data = json.load(f)
          
                    for item in data["data"]["transactionData"]:
                        name = item["name"]
                        count = item["paymentInstruments"][0]["count"]
                        amount = item["paymentInstruments"][0]["amount"]

                        clm['State'].append(state)
                        clm['Year'].append(year)
                        clm['Quarter'].append(int(q.strip(".json")))
                        clm['Transaction_type'].append(name)
                        clm['Transaction_count'].append(count)
                        clm['Transaction_amount'].append(amount)


Agg_Trans= pd.DataFrame(clm)
Agg_Trans
```

```python
import json
import pandas as pd
import os
path ="C:\\Users\\yadun\\OneDrive\\Desktop\\Abhinand\\Project\\pulse\\data\\top\\insurance\\country\\india\\state"

Top_state_list=os.listdir(path)


clm={
    'state':[],
    'year':[],
    'quarter':[],
    'Insurance_type':[],
    'Insurance_count':[],
    'Insurance_amount':[],
    'Pincode':[]
}
for state in Top_state_list:
    state_path=os.path.join(path,state)
    if not os.path.isdir(state_path):                      
        continue
    year_list = os.listdir(state_path)

    for year in year_list:
        year_path=os.path.join(state_path,year)
        if not os.path.isdir(year_path):
            continue
        quarter_list=os.listdir(year_path)

        for quarter in quarter_list:
            if quarter.endswith(".json"):
                
                json_file_path=os.path.join(year_path,quarter)
                
                with open(json_file_path, "r") as f:
                    data = json.load(f)
                for z in data["data"].get("district", []):
                    
                    clm['Insurance_type'].append(z["entityName"])
                    clm['Insurance_count'].append(z["metric"]["count"])
                    clm['Insurance_amount'].append(z["metric"]["amount"])
                    clm['Pincode'].append(None)

                    clm['state'].append(state)
                    clm['year'].append(year)
                    clm['quater'].append(int(quarter.replace(".json", "")))

                for z in data["data"].get("pincodes", []):
                    clm['Insurance_type'].append(None)
                    clm['Insurance_count'].append(z["metric"]["count"])
                    clm['Insurance_amount'].append(z["metric"]["amount"])
                    clm['Pincode'].append(z["entityName"])

                    clm['state'].append(state)
                    clm['year'].append(year)
                    clm['quarter'].append(int(quarter.replace(".json", "")))
Top_Insurance=pd.DataFrame(clm)
Top_Insurance   
```
